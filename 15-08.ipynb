```python
numbers = []
cutler = []
tekler = []

while True:
    eded = int(input("Eded daxil et (dayanmaq ucun 0): "))

    if eded == 0:
        break

    numbers.append(eded)

    if eded % 2 == 0:
        cutler.append(eded)
    else:
        tekler.append(eded)

print("Butun ededler:", numbers)

print("Cut ededler:", cutler)
print("Cut ededlerin sayi:", len(cutler))
print("Cut ededlerin cemi:", sum(cutler))

print("Tek ededler:", tekler)
print("Tek ededlerin sayi:", len(tekler))
print("Tek ededlerin cemi:", sum(tekler))
```

```python
numbers = []
menfi = []
musbet = []

for i in range(5):
    eded = int(input("Ededi daxil edin: "))

    if eded == 0:
        break

    if eded < 0:
        menfi.append(eded)
        continue

    numbers.append(eded)
    musbet.append(eded)

print("Numbers:", numbers)
print("Menfi:", menfi)
print("Musbet:", musbet)
    
```
