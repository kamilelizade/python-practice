```python
# 1-dən 30-a qədər ədədləri yoxla:

# 3-ə bölünürsə → "3-e bolunur"
# 5-ə bölünürsə → "5-e bolunur"
# həm 3-ə, həm 5-ə bölünürsə → "Hem 3-e, hem 5-e bolunur"
# heç birinə bölünmürsə → heç nə çap etmə.


for i in range(1,31):
    if i % 3 == 0:
        print(i,"Bu eded 3e bolunur")
    if i % 5 == 0:    
        print(i,"Bu eded 5e bolunur")
    if i % 3 == 0 and i % 5 == 0:
        print(i, "Bu eded hem 3e, hem de 5e bolunur")

        
        
```

```python
# 1-dən 100-ə qədər:

# cüt ədədlərin cəmini tap
# tək ədədlərin cəmini tap
# sonda hər ikisini çap et

# Nəticə belə olmalıdır:

# Cut ededlerin cemi: ...
# Tek ededlerin cemi: ...

cut_total = 0
tek_total = 0
for i in range(1,101):
    if i % 2 == 0:
      cut_total+=i
    else:
        tek_total+=i
        
print("Cut ededlerin cemi:", cut_total)
print("Tek ededlerin cemi:", tek_total)
    
        
```

```python
#İstifadəçidən 10 ədəd al.

#Proqram:

#Hər ədədi qəbul etsin.
#Cütlərin sayını tapsın.
#Təkların sayını tapsın.
#Cütlərin cəmini tapsın.
#Təkların cəmini tapsın.
#Sonda bütün nəticələri göstərin

cut_total = 0
tek_total = 0

cut_count = 0
tek_count = 0

for i in range(10):
    number = int(input("Eded daxil et: "))

    if number % 2 == 0:
        cut_total += number
        cut_count += 1
    else:
        tek_total += number
        tek_count += 1

print("Cut ededlerin cemi:", cut_total)
print("Tek ededlerin cemi:", tek_total)
print("Cut ededlerin sayi:", cut_count)
print("Tek ededlerin sayi:", tek_count)
```
